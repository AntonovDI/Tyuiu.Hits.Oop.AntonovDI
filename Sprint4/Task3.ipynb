```
public interface IAnimal
{
    void MakeSound();
    void Move();
    void Eat(string food);
    void Sleep();
    int GetNumberOfLegs();
    string GetFullInfo();
}

public interface IPet : IAnimal
{
    void Play();
    void VisitVet();
}

public abstract class Animal : IAnimal
{
    public string Name { get; set; }
    public double Weight { get; set; }
    public string Habitat { get; set; }

    private int _age;
    public int Age
    {
        get { return _age; }
        set
        {
            if (value >= 0) _age = value;
        }
    }

    public abstract void MakeSound();
    public abstract int GetNumberOfLegs();

    void IAnimal.Move()
    {
        Console.WriteLine($"{Name} передвигается.");
    }

    void IAnimal.Eat(string food)
    {
        Console.WriteLine($"{Name} ест {food}.");
    }

    public virtual void Sleep()
    {
        Console.WriteLine($"{Name} спит.");
    }

    public virtual string GetFullInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Вес: {Weight} кг, Среда: {Habitat}";
    }
}

public class Dog : Animal, IPet, IAnimal
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    public string FavoriteToy { get; set; }
    public string CoatColor { get; set; }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Гав-гав!");
    }

    public override int GetNumberOfLegs()
    {
        return 4;
    }

    void IAnimal.Move()
    {
        Console.WriteLine($"{Name} бежит на {GetNumberOfLegs()} лапах.");
    }

    void IAnimal.Eat(string food)
    {
        Console.WriteLine($"{Name} грызет {food}.");
    }

    void IPet.Play()
    {
        Console.WriteLine($"{Name} играет с игрушкой: {FavoriteToy}.");
    }

    void IPet.VisitVet()
    {
        Console.WriteLine($"{Name} проходит осмотр у ветеринара.");
    }

    public void Guard()
    {
        Console.WriteLine($"{Name} охраняет территорию.");
    }

    public void ExecuteCommand(string command)
    {
        Console.WriteLine(IsTrained ? $"{Name} выполняет команду '{command}'." : $"{Name} не обучен команде '{command}'.");
    }

    public override string GetFullInfo()
    {
        return $"{base.GetFullInfo()}, Порода: {Breed}, Окрас: {CoatColor}";
    }
}

public class Cat : Animal, IPet, IAnimal
{
    public bool IsIndoor { get; set; }
    public string EyeColor { get; set; }
    public int LivesLeft { get; set; }
    public bool HasClaws { get; set; }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Мяу!");
    }

    public override int GetNumberOfLegs()
    {
        return 4;
    }

    void IAnimal.Move()
    {
        Console.WriteLine($"{Name} бесшумно крадется.");
    }

    void IAnimal.Eat(string food)
    {
        Console.WriteLine($"{Name} аккуратно ест {food}.");
    }

    void IPet.Play()
    {
        Console.WriteLine($"{Name} гоняется за лазерной указкой.");
    }

    void IPet.VisitVet()
    {
        Console.WriteLine($"{Name} получает прививку у ветеринара.");
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} громко мурлычет.");
    }

    public void Scratch()
    {
        Console.WriteLine(HasClaws ? $"{Name} точит когти." : $"{Name} не может точить когти.");
    }

    public override string GetFullInfo()
    {
        return $"{base.GetFullInfo()}, Цвет глаз: {EyeColor}, Домашняя: {IsIndoor}";
    }
}

public class Bird : Animal, IAnimal
{
    public double WingSpan { get; set; }
    public string FeatherColor { get; set; }
    public bool CanFly { get; set; }
    public string BeakType { get; set; }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} поет: Чирик-чирик!");
    }

    public override int GetNumberOfLegs()
    {
        return 2;
    }

    void IAnimal.Move()
    {
        Console.WriteLine(CanFly ? $"{Name} летит (размах крыльев {WingSpan} м)." : $"{Name} прыгает по земле.");
    }

    void IAnimal.Eat(string food)
    {
        Console.WriteLine($"{Name} клюет {food}.");
    }

    public void BuildNest()
    {
        Console.WriteLine($"{Name} вьет гнездо.");
    }

    public void CleanFeathers()
    {
        Console.WriteLine($"{Name} чистит перья ({FeatherColor}).");
    }

    public override string GetFullInfo()
    {
        return $"{base.GetFullInfo()}, Оперение: {FeatherColor}, Клюв: {BeakType}";
    }
}

public class AnimalCareService
{
    private readonly IAnimal _animal;

    public AnimalCareService(IAnimal animal)
    {
        _animal = animal;
    }

    public void DailyRoutine(string food)
    {
        Console.WriteLine(_animal.GetFullInfo());
        _animal.MakeSound();
        _animal.Move();
        _animal.Eat(food);
        _animal.Sleep();
    }
}

public class PetCareService
{
    private readonly IPet _pet;

    public PetCareService(IPet pet)
    {
        _pet = pet;
    }

    public void PetRoutine(string food)
    {
        Console.WriteLine(_pet.GetFullInfo());
        _pet.MakeSound();
        _pet.Eat(food);
        _pet.Play();
        _pet.VisitVet();
    }
}

Dog dog = new Dog
{
    Name = "Рекс",
    Age = 4,
    Weight = 28.5,
    Habitat = "Дом",
    Breed = "Немецкая овчарка",
    IsTrained = true,
    FavoriteToy = "Мяч",
    CoatColor = "Чепрачный"
};

Cat cat = new Cat
{
    Name = "Мурка",
    Age = 2,
    Weight = 4.2,
    Habitat = "Квартира",
    IsIndoor = true,
    EyeColor = "Зеленый",
    LivesLeft = 9,
    HasClaws = true
};

Bird bird = new Bird
{
    Name = "Кеша",
    Age = 1,
    Weight = 0.1,
    Habitat = "Клетка",
    WingSpan = 0.25,
    FeatherColor = "Синий",
    CanFly = true,
    BeakType = "Крючковатый"
};

PetCareService dogService = new PetCareService(dog);
dogService.PetRoutine("корм");
dog.Guard();
dog.ExecuteCommand("Сидеть");

Console.WriteLine();

PetCareService catService = new PetCareService(cat);
catService.PetRoutine("рыбу");
cat.Purr();
cat.Scratch();

Console.WriteLine();

AnimalCareService birdService = new AnimalCareService(bird);
birdService.DailyRoutine("зерно");
bird.BuildNest();
bird.CleanFeathers();
```
