<h2 style="color:DodgerBlue">Название проекта:</h2>

----

### Вариант задания: Автоматизированная система учета обитателей цифрового заповедника (Digital Shelter Management System)


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект представляет собой объектно-ориентированную модель информационной системы для управления экосистемой цифрового заповедника или приюта для животных. Система позволяет регистрировать различные биологические виды, отслеживать их параметры, симулировать их жизненные циклы (передвижение, сон, питание, звуковые сигналы) и управлять группами животных с использованием безопасных обобщенных контейнеров.

#### Дополнительное задание
В архитектуру были внесены следующие расширения:
1. **Базовый класс `Animal`** расширен 4 атрибутами (`Name`, `Age`, `Weight`, `Habitat`) и 4 группами методов (`Move`, `MakeSound`, `Eat`, `Sleep`).
2. **Реализован полиморфизм и перекрытие методов (Overriding)**: каждый дочерний класс (`Dog`, `Cat`, `Bird`) переопределяет базовое поведение виртуальных методов под свои биологические особенности.
3. **Реализована перегрузка методов (Overloading)**: метод `Move` имеет три разные сигнатуры для поддержки различных контекстов симуляции (без параметров, с указанием скорости, с указанием цели и длительности).
4. **Разработан Generic-класс `AnimalShelter<T>`**: обобщенный контейнер с ограничением типа (`where T : Animal`), реализующий строго типизированное управление коллекциями обитателей заповедника.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public class Animal
{
    private string name;
    private int age;
    private double weight;
    private string habitat;

    public string Name
    {
        get { return name; }
        set { if (!string.IsNullOrWhiteSpace(value)) name = value; }
    }

    public int Age
    {
        get { return age; }
        set { if (value >= 0) age = value; }
    }

    public double Weight
    {
        get { return weight; }
        set { if (value > 0) weight = value; }
    }

    public string Habitat
    {
        get { return habitat; }
        set { if (!string.IsNullOrWhiteSpace(value)) habitat = value; }
    }

    public Animal(string name, int age, double weight, string habitat)
    {
        Name = name;
        Age = age;
        Weight = weight;
        Habitat = habitat;
    }

    public virtual void Move()
    {
        Console.WriteLine($"{Name} передвигается в среде обитания: {Habitat}.");
    }

    public virtual void Move(int speed)
    {
        Console.WriteLine($"{Name} движется со скоростью {speed} км/ч.");
    }

    public virtual void Move(string destination, int duration)
    {
        Console.WriteLine($"{Name} направляется в {destination} в течение {duration} мин.");
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает неопределенный звук.");
    }

    public virtual void Eat(string food)
    {
        Console.WriteLine($"{Name} ест {food}.");
    }

    public virtual void Sleep(int hours)
    {
        Console.WriteLine($"{Name} спит в течение {hours} ч.");
    }
}

public class Dog : Animal
{
    private string breed;

    public string Breed
    {
        get { return breed; }
        set { if (!string.IsNullOrWhiteSpace(value)) breed = value; }
    }

    public Dog(string name, int age, double weight, string habitat, string breed) 
        : base(name, age, weight, habitat)
    {
        Breed = breed;
    }

    public override void Move()
    {
        Console.WriteLine($"Собака {Name} породы {Breed} бегает на четырех лапах.");
    }

    public override void Move(int speed)
    {
        base.Move(speed);
        Console.WriteLine($"{Name} активно виляет хвостом при быстром беге.");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"Собака {Name} громко лает: Гав-гав!");
    }

    public override void Eat(string food)
    {
        Console.WriteLine($"Собака {Name} с аппетитом грызет {food}.");
    }

    public override void Sleep(int hours)
    {
        Console.WriteLine($"Собака {Name} свернулась калачиком и спит {hours} ч.");
    }
}

public class Cat : Animal
{
    private bool isLazy;

    public bool IsLazy
    {
        get { return isLazy; }
        set { isLazy = value; }
    }

    public Cat(string name, int age, double weight, string habitat, bool isLazy) 
        : base(name, age, weight, habitat)
    {
        IsLazy = isLazy;
    }

    public override void Move()
    {
        Console.WriteLine($"Кошка {Name} бесшумно крадется по территории.");
    }

    public override void Move(string destination, int duration)
    {
        Console.WriteLine($"Кошка {Name} запрыгивает на {destination} за {duration} сек.");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"Кошка {Name} нежно мурлычет: Мяу-мур.");
    }

    public override void Eat(string food)
    {
        Console.WriteLine($"Кошка {Name} аккуратно лакает {food}.");
    }

    public override void Sleep(int hours)
    {
        string status = IsLazy ? "лениво дремлет" : "спит";
        Console.WriteLine($"Кошка {Name} {status} {hours} ч. на солнце.");
    }
}

public class Bird : Animal
{
    private double wingspan;

    public double Wingspan
    {
        get { return wingspan; }
        set { if (value > 0) wingspan = value; }
    }

    public Bird(string name, int age, double weight, string habitat, double wingspan) 
        : base(name, age, weight, habitat)
    {
        Wingspan = wingspan;
    }

    public override void Move()
    {
        Console.WriteLine($"Птица {Name} парит в небе с размахом крыльев {Wingspan} м.");
    }

    public override void Move(int speed)
    {
        Console.WriteLine($"Птица {Name} летит со скоростью {speed} км/ч, ловя потоки ветра.");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"Птица {Name} звонко поет: Чирик-чирик!");
    }

    public override void Eat(string food)
    {
        Console.WriteLine($"Птица {Name} быстро клюет {food}.");
    }

    public override void Sleep(int hours)
    {
        Console.WriteLine($"Птица {Name} спрятала голову под крыло и спит {hours} ч.");
    }
}

public class AnimalShelter<T> where T : Animal
{
    private List<T> _animals = new List<T>();

    public void Add(T animal)
    {
        _animals.Add(animal);
    }

    public void Remove(T animal)
    {
        _animals.Remove(animal);
    }

    public void ManageShelter()
    {
        foreach (var animal in _animals)
        {
            Console.WriteLine($"--- Управление животным: {animal.Name} ---");
            animal.Move();
            animal.MakeSound();
            animal.Eat("корм");
            animal.Sleep(8);
            Console.WriteLine();
        }
    }
}

AnimalShelter<Animal> ecosystem = new AnimalShelter<Animal>();
ecosystem.Add(new Dog("Бобик", 3, 12.5, "Дом", "Овчарка"));
ecosystem.Add(new Cat("Мурка", 2, 4.2, "Квартира", true));
ecosystem.Add(new Bird("Кеша", 1, 0.3, "Лес", 0.5));

ecosystem.ManageShelter();

Console.WriteLine("--- Демонстрация перегруженных методов вне коллекции ---");
Dog testDog = new Dog("Рекс", 4, 15.0, "Двор", "Лабрадор");
testDog.Move(20);

Cat testCat = new Cat("Барсик", 5, 5.0, "Дом", false);
testCat.Move("шкаф", 2);
```
