### Вариант задания 


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----
Создать базовый класс Employee в C#, который будет представлять информацию о
сотрудниках компании. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.
Требования к базовому классу Employee:
• Атрибуты: Имя (Name), Возраст (Age), Зарплата (Salary).
• Методы:
o GetInfo(): метод для получения информации о сотруднике в виде
строки.
o Work(): метод для вывода сообщения о выполнении рабочих
обязанностей.
o TakeVacation(): метод для вывода сообщения о взятии отпуска.
Требования к производным классам:
1. Менеджер (Manager): Должен содержать дополнительные атрибуты, такие
как Количество подчиненных (SubordinatesCount). Метод Work() должен быть
переопределен для добавления информации о управлении командой.
2. Разработчик (Developer): Должен содержать дополнительные атрибуты,
такие как Специализация (Specialization). Метод TakeVacation() должен быть
переопределен для добавления информации о необходимости согласования
отпуска с проектными сроками.
3. Дизайнер (Designer) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Стиль дизайна (DesignStyle).
Метод Work() должен быть переопределен для добавления информации о
работе над дизайн-проектами.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
public interface IEvaluatable
{
    int PerformanceScore { get; set; }
    void EvaluatePerformance();
}

public interface IProjectTrackable
{
    string CurrentProject { get; set; }
    void UpdateProjectStatus(string status);
}


public class Employee
{
    private string _name;
    private int _age;
    private decimal _salary;
    private int _experience;

    public string Name
    {
        get => _name;
        set => _name = !string.IsNullOrWhiteSpace(value) ? value : throw new ArgumentException("Имя не может быть пустым.");
    }

    public int Age
    {
        get => _age;
        set => _age = value >= 18 ? value : throw new ArgumentException("Возраст сотрудника должен быть не менее 18 лет.");
    }

    public decimal Salary
    {
        get => _salary;
        set => _salary = value >= 0 ? value : throw new ArgumentException("Зарплата не может быть отрицательной.");
    }

    public int Id { get; set; }
    public string Department { get; set; }
    
    public int Experience
    {
        get => _experience;
        set => _experience = value >= 0 ? value : throw new ArgumentException("Стаж не может быть отрицательным.");
    }


    public Employee(int id, string name, int age, decimal salary, string department, int experience)
    {
        Id = id;
        Name = name;
        Age = age;
        Salary = salary;
        Department = department;
        Experience = experience;
    }

    public virtual string GetInfo()
    {
        return $"[ID: {Id}] {Name} | Возраст: {Age} | Стаж: {Experience} г. | Департамент: {Department} | Зарплата: {Salary:C}";
    }

    public virtual void Work()
    {
        Console.WriteLine($"{Name} выполняет стандартные должностные обязанности.");
    }

    public virtual void TakeVacation()
    {
        Console.WriteLine($"{Name} подал заявление на плановый отпуск.");
    }
}


public class Developer : Employee, IProjectTrackable, IEvaluatable
{
    private string _currentProject;
    
    public string Specialization { get; set; }
    public string ProgrammingLanguage { get; set; }
    public string GitHubProfile { get; set; }

    public string CurrentProject
    {
        get => _currentProject;
        set => _currentProject = !string.IsNullOrWhiteSpace(value) ? value : "Бэклог";
    }
    public int PerformanceScore { get; set; }

    public Developer(int id, string name, int age, decimal salary, string department, int experience, 
                     string specialization, string lang, string github) 
        : base(id, name, age, salary, department, experience)
    {
        Specialization = specialization;
        ProgrammingLanguage = lang;
        GitHubProfile = github;
        CurrentProject = "Без проекта";
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} пишет код на {ProgrammingLanguage} под проект '{CurrentProject}'.");
    }

    public override void TakeVacation()
    {
        Console.WriteLine($"{Name} ({Specialization}) проверяет дедлайны по проекту '{CurrentProject}' перед согласованием отпуска.");
    }

    public void UpdateProjectStatus(string status) => Console.WriteLine($"[Проект {CurrentProject}]: Статус обновлен разработчиком {Name} на '{status}'.");
    public void EvaluatePerformance() => Console.WriteLine($"Оценка эффективности разработчика {Name}: {PerformanceScore}/100");
}

public class Designer : Employee
{
    public string DesignStyle { get; set; }
    public string PortfolioUrl { get; set; }
    public string ToolUsed { get; set; } 

    public Designer(int id, string name, int age, decimal salary, string department, int experience, 
                    string style, string portfolio, string tool) 
        : base(id, name, age, salary, department, experience)
    {
        DesignStyle = style;
        PortfolioUrl = portfolio;
        ToolUsed = tool;
    }

    public override string GetInfo() => base.GetInfo() + $" | Стиль: {DesignStyle} | Инструмент: {ToolUsed}";

    public override void Work()
    {
        Console.WriteLine($"{Name} отрисовывает интерактивные макеты в {ToolUsed} (Стиль: {DesignStyle}).");
    }
}

public class Manager : Employee
{
    public int SubordinatesCount { get; set; }
    public string BudgetCode { get; set; }
    public string ManagementMethodology { get; set; } 

    public Manager(int id, string name, int age, decimal salary, string department, int experience, 
                   int subordinatesCount, string budgetCode, string methodology) 
        : base(id, name, age, salary, department, experience)
    {
        SubordinatesCount = subordinatesCount;
        BudgetCode = budgetCode;
        ManagementMethodology = methodology;
    }

    public override void Work()
    {
        Console.WriteLine($"{Name} координирует команду из {SubordinatesCount} человек по методологии {ManagementMethodology}.");
    }

    // Взаимодействие объектов между собой
    public void AssignTaskToDeveloper(Developer dev, string taskName)
    {
        Console.WriteLine($"[Взаимодействие] Менеджер {Name} назначает задачу '{taskName}' разработчику {dev.Name}.");
        dev.CurrentProject = "Новая Платформа";
    }
}

public class TeamLead : Manager
{
    public string ArchitectureRole { get; set; }
    public int ApprovedArchitecturesCount { get; set; }

    public TeamLead(int id, string name, int age, decimal salary, string department, int experience, 
                    int subordinatesCount, string budgetCode, string methodology, string archRole) 
        : base(id, name, age, salary, department, experience, subordinatesCount, budgetCode, methodology)
    {
        ArchitectureRole = archRole;
        ApprovedArchitecturesCount = 0;
    }

    public override void Work()
    {
        base.Work(); 
        Console.WriteLine($"{Name} дополнительно проводит код-ревью и выполняет роль: {ArchitectureRole}.");
    }
}
Developer dev = new Developer(101, "Артем Сидоров", 25, 140000m, "R&D", 3, "Backend", "C#", "://github.com");
Designer des = new Designer(102, "Алена Игорева", 29, 110000m, "Design", 5, "UI/UX Минимализм", "behance.net/alyona", "Figma");
Manager mgr = new Manager(501, "Дмитрий Николаевич", 42, 200000m, "Management", 12, 10, "B-2026", "Scrum");
TeamLead lead = new TeamLead(601, "Валерий Петров", 35, 250000m, "R&D", 8, 5, "TL-99", "Agile/Safe", "Solution Architect");

Console.WriteLine("================ Вывод информации об объектах ================");
Console.WriteLine(dev.GetInfo());
Console.WriteLine(des.GetInfo());
Console.WriteLine(mgr.GetInfo());
Console.WriteLine(lead.GetInfo());

Console.WriteLine("\n================ Проверка Полиморфизма (Work) ================");
dev.Work();
des.Work();
mgr.Work();
lead.Work(); 

Console.WriteLine("\n================ Проверка Множественного наследования ================");
dev.PerformanceScore = 92;
dev.EvaluatePerformance();
dev.UpdateProjectStatus("В разработке");

Console.WriteLine("\n================ Взаимодействие объектов ================");

mgr.AssignTaskToDeveloper(dev, "Интеграция платежного шлюза");
dev.Work(); 
```
