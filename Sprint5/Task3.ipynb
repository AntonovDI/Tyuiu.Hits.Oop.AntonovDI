```
using System;
using System.Collections.Generic;
using System.Linq;

public delegate void AnimalRoutineDelegate();

public class Animal
{
    public string Id { get; set; }
    public string Name { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }
    public bool IsHungry { get; set; }

    public delegate void AnimalStatusHandler(string message);
    public event AnimalStatusHandler StatusChanged;

    public Animal(string id, string name, int age, double weight)
    {
        Id = id;
        Name = name;
        Age = age;
        Weight = weight;
        IsHungry = true;
    }

    public void Feed(double foodWeight)
    {
        IsHungry = false;
        Weight += foodWeight * 0.1;
        StatusChanged?.Invoke($"[{Id}] {Name} покормлен(а). Текущий вес: {Weight:F1} кг.");
    }

    public void CelebrateBirthday()
    {
        Age++;
        StatusChanged?.Invoke($"[{Id}] {Name} отмечает день рождения! Новый возраст: {Age}.");
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издает звук.");
    }

    public virtual void DisplayInfo()
    {
        Console.WriteLine($"ID: {Id} | Имя: {Name} | Возраст: {Age} | Вес: {Weight:F1} кг | Голоден: {IsHungry}");
    }

    public override bool Equals(object obj)
    {
        if (obj is Animal other)
        {
            return Id == other.Id;
        }
        return false;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Id);
    }
}

public class Dog : Animal
{
    public string Breed { get; set; }
    public bool IsTrained { get; set; }
    public string FavoriteToy { get; set; }
    public int WalkCount { get; set; }

    public Dog(string id, string name, int age, double weight, string breed, bool isTrained, string favoriteToy)
        : base(id, name, age, weight)
    {
        Breed = breed;
        IsTrained = isTrained;
        FavoriteToy = favoriteToy;
        WalkCount = 0;
    }

    public void Walk()
    {
        WalkCount++;
        IsHungry = true;
        Console.WriteLine($"{Name} погулял(а). Всего прогулок: {WalkCount}.");
    }

    public void Train()
    {
        IsTrained = true;
        Console.WriteLine($"{Name} успешно прошел(ла) дрессировку.");
    }

    public void Play()
    {
        Console.WriteLine($"{Name} играет с игрушкой: {FavoriteToy}.");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Гав-гав!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Порода: {Breed} | Дрессирован: {IsTrained} | Игрушка: {FavoriteToy} | Прогулок: {WalkCount}");
    }
}

public class Cat : Animal
{
    public bool IsIndoor { get; set; }
    public string FurColor { get; set; }
    public int LivesLeft { get; set; }
    public int SleepHours { get; set; }

    public Cat(string id, string name, int age, double weight, bool isIndoor, string furColor)
        : base(id, name, age, weight)
    {
        IsIndoor = isIndoor;
        FurColor = furColor;
        LivesLeft = 9;
        SleepHours = 0;
    }

    public void Sleep(int hours)
    {
        SleepHours += hours;
        Console.WriteLine($"{Name} поспал(а) {hours} ч. Всего сна: {SleepHours} ч.");
    }

    public void Purr()
    {
        Console.WriteLine($"{Name} громко мурлычет.");
    }

    public void Scratch()
    {
        Console.WriteLine($"{Name} точит когти.");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Мяу!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Домашняя: {IsIndoor} | Окрас: {FurColor} | Жизней: {LivesLeft} | Сон: {SleepHours} ч.");
    }
}

public class Bird : Animal
{
    public bool CanFly { get; set; }
    public double Wingspan { get; set; }
    public string FeatherColor { get; set; }
    public bool CanTalk { get; set; }

    public Bird(string id, string name, int age, double weight, bool canFly, double wingspan, string featherColor, bool canTalk)
        : base(id, name, age, weight)
    {
        CanFly = canFly;
        Wingspan = wingspan;
        FeatherColor = featherColor;
        CanTalk = canTalk;
    }

    public void Fly()
    {
        if (CanFly)
            Console.WriteLine($"{Name} летает по вольеру.");
        else
            Console.WriteLine($"{Name} не умеет летать.");
    }

    public void Sing()
    {
        Console.WriteLine($"{Name} поет мелодичную песню.");
    }

    public void RepeatPhrase(string phrase)
    {
        if (CanTalk)
            Console.WriteLine($"{Name} повторяет: \"{phrase}\"");
        else
            Console.WriteLine($"{Name} не умеет говорить.");
    }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} говорит: Чирик-чирик!");
    }

    public override void DisplayInfo()
    {
        base.DisplayInfo();
        Console.WriteLine($"Летает: {CanFly} | Размах крыльев: {Wingspan} м | Оперение: {FeatherColor} | Говорит: {CanTalk}");
    }
}

Dog dog = new Dog("D01", "Рекс", 4, 28.5, "Овчарка", true, "Мяч");
Cat cat = new Cat("C01", "Мурка", 2, 4.2, true, "Рыжий");
Bird bird = new Bird("B01", "Кеша", 3, 0.4, true, 0.35, "Зеленый", true);

dog.StatusChanged += msg => Console.WriteLine($"[УВЕДОМЛЕНИЕ] {msg}");
cat.StatusChanged += msg => Console.WriteLine($"[УВЕДОМЛЕНИЕ] {msg}");
bird.StatusChanged += msg => Console.WriteLine($"[УВЕДОМЛЕНИЕ] {msg}");

List<Animal> zooList = new List<Animal> { dog, cat, bird };
Dictionary<string, Animal> zooCatalog = new Dictionary<string, Animal>();
HashSet<Animal> uniqueAnimals = new HashSet<Animal>();

foreach (var animal in zooList)
{
    zooCatalog[animal.Id] = animal;
    uniqueAnimals.Add(animal);
}

AnimalRoutineDelegate morningRoutine = dog.Walk;
morningRoutine += () => cat.Sleep(3);
morningRoutine += bird.Fly;
morningRoutine();

Console.WriteLine();

dog.Feed(1.5);
cat.Feed(0.3);
bird.CelebrateBirthday();

Console.WriteLine();

if (zooCatalog.TryGetValue("B01", out Animal foundAnimal) && foundAnimal is Bird parrot)
{
    parrot.RepeatPhrase("Привет, мир!");
}

Console.WriteLine("\n--- Список всех животных в зоопарке ---");
var sortedAnimals = uniqueAnimals.OrderBy(a => a.Age);
foreach (var animal in sortedAnimals)
{
    animal.DisplayInfo();
    animal.MakeSound();
    Console.WriteLine();
}
```
